```python
# Linked List implementation of a playlist
class Song:
    def __init__(self, title, artist, duration):
        self.title = title
        self.artist = artist
        self.duration = duration
        self.next = None

class Playlist:
    def __init__(self):
        self.head = None
        self.current_playing = None

    def add_song(self, title, artist, duration):
        new_song = Song(title, artist, duration)
        if not self.head:
            self.head = new_song
            self.current_playing = new_song
        else:
            current = self.head
            while current.next:
                current = current.next
            current.next = new_song

    def play_next(self):
        if self.current_playing and self.current_playing.next:
            self.current_playing = self.current_playing.next
            return f"Now playing: {self.current_playing.title} by {self.current_playing.artist}"
        return "End of playlist"

    def display_playlist(self):
        current = self.head
        while current:
            print(f"{current.title} - {current.artist} ({current.duration})")
            current = current.next
```

```python
# sorting algorithms class

class Sorting:
    
    def __init__(self):
        self.arr = None
        
    def bubble_sort(self, arr):
        n = len(arr)
        for i in range(n-1):
            for j in range(n-i-1):
                if arr[j] > arr[j + 1]:
                    arr[j], arr[j+1] = arr[j+1], arr[j]
        return arr
    
    def merge(self, l, r):
        sorted_list = []
        i = j = 0
        
        while i < len(l) and j < len(r):
            if l[i] < r[j]:
                sorted_list.append(l[i])
                i += 1
            else:
                sorted_list.append(r[j])
                j += 1
                
        sorted_list.extend(l[i:]), sorted_list.extend(r[j:])
        
        return sorted_list
    
    def merge_sort(self,arr):
        if len(arr) <=  1:
            return arr
        
        mid = len(arr) // 2
        left = arr[:mid]
        right = arr[mid:]
        
        left_merge = self.merge_sort(left)
        right_merge = self.merge_sort(right)
        
        return self.merge(left_merge, right_merge)
    
    
obj = Sorting()
obj.bubble_sort([3,2,5,1])
obj.merge_sort([3,2,5,1])
    
        
```

```python
# Binary search tree

class BinaryNode:
    def __init__(self, value):
        self.left = None
        self.right = None
        self.val = value
        
    def insert(self, value):
        if self.val:
            if value < self.val:
                if self.left is None:
                    self.left = BinaryNode(value)
                else:
                    self.left.insert(value)
            elif value > self.val:
                if not self.right:
                    self.right = BinaryNode(value)
                else:
                    self.right.insert(value)
                    
        else:
            self.val = value
            
            
            
obj = BinaryNode(10)
obj.insert(5)
obj.insert(6)
obj.insert(3)

            
```
